## 1. O que são Estruturas de Repetição?

**Estruturas de repetição** (também chamadas de **laços** ou **loops**) permitem executar um bloco de código várias vezes, sem precisar escrever o mesmo código repetidamente.

Pense em situações do dia a dia:
- Contar de 1 até 10
- Imprimir os nomes de todos os alunos
- Repetir uma ação até que uma condição seja atendida
- Processar uma lista de itens

Em Python, temos dois tipos principais de laços:
- **`while`**: repete enquanto uma condição for verdadeira
- **`for`**: repete para cada item de uma sequência


## 2. Laço while (Enquanto)

O laço `while` repete um bloco de código **enquanto** uma condição for verdadeira.

**Sintaxe:**
```python
while condição:
    # código que será executado enquanto a condição for True
```

**Importante:** A condição é verificada **antes** de cada iteração. Se for `False` desde o início, o código dentro do `while` nunca será executado!

### 2.1 Cuidado: Loops Infinitos!

Um **loop infinito** acontece quando a condição do `while` nunca se torna `False`. Isso faz o programa rodar para sempre!

**Exemplo de loop infinito (NÃO execute por muito tempo!):**
```python
contador = 1
while contador <= 5:  # contador nunca muda!
    print(contador)
    # Falta: contador = contador + 1
```

**Sempre certifique-se de que a condição do `while` eventualmente se tornará `False`!**

## 4. Comparando while e for

### Quando usar `while`?
- Quando não sabemos quantas vezes vamos repetir
- Quando queremos repetir até que uma condição seja atendida
- Quando a condição depende de algo que muda durante a execução

### Quando usar `for`?
- Quando sabemos quantos itens vamos processar
- Quando queremos iterar sobre uma sequência (lista, string, etc.)
- Quando queremos processar cada item de uma coleção

**Dica:** Na maioria dos casos, `for` é mais seguro e mais fácil de usar!

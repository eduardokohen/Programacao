## 1. Função range()

A função `range()` gera uma sequência de números. É muito útil com o laço `for` quando queremos repetir algo um número específico de vezes.

**Sintaxes:**
```python
range(fim)              # De 0 até fim-1
range(inicio, fim)      # De inicio até fim-1
range(inicio, fim, passo)  # De inicio até fim-1, pulando de passo em passo
```

**Importante:** `range()` não cria uma lista diretamente. Ela cria um objeto que gera números sob demanda (mais eficiente).

| `continue` | Pula para a próxima iteração | Quando queremos ignorar alguns itens |
| `pass` | Não faz nada (placeholder) | Quando precisamos de sintaxe válida |

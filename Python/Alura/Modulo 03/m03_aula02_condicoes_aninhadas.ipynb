## 1. O que são Condições Aninhadas?

**Condições aninhadas** são estruturas condicionais dentro de outras estruturas condicionais. É como ter decisões dentro de decisões!

Pense em situações do dia a dia:
- **Se** estiver chovendo:
  - **Se** tiver guarda-chuva, **então** saia
  - **Senão**, fique em casa
- **Se** aprovado:
  - **Se** nota >= 9, **então** "Parabéns!"
  - **Senão**, "Bom trabalho!"

Em Python, podemos colocar `if`, `elif` e `else` dentro de outros `if`, `elif` e `else`!

## 5. Quando Usar Condições Aninhadas?

Condições aninhadas são úteis quando:
- Precisamos verificar condições que dependem de outras condições
- Queremos organizar decisões em níveis lógicos
- A lógica do problema tem uma estrutura hierárquica

**Cuidado:** Evite aninhar mais de 3-4 níveis, pois o código pode ficar difícil de ler!

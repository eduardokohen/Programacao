## 1. O que são Estruturas Condicionais?

**Estruturas condicionais** permitem que o programa tome decisões baseadas em condições. É como dizer ao computador: "Se isso acontecer, faça aquilo. Caso contrário, faça outra coisa."

Pense em situações do dia a dia:
- **Se** estiver chovendo, **então** leve o guarda-chuva
- **Se** a nota for maior que 7, **então** aprovado
- **Se** tiver dinheiro, **então** compre o produto

Em Python, usamos `if`, `elif` e `else` para criar essas estruturas!

## 4. Estrutura if-elif-else (Se-Senão Se-Senão)

Quando temos **múltiplas condições** para verificar, usamos `elif` (abreviação de "else if").

**Sintaxe:**
```python
if condição1:
    # código se condição1 for True
elif condição2:
    # código se condição2 for True
elif condição3:
    # código se condição3 for True
else:
    # código se nenhuma condição for True
```

**Importante:** O Python verifica as condições **na ordem** e executa apenas o **primeiro** bloco que for verdadeiro!

## 7. Resumo da Aula

Nesta aula você aprendeu:

**Estrutura `if`:**
- Executa código apenas se a condição for `True`
- Sintaxe: `if condição:`

**Estrutura `if-else`:**
- Executa um código se `True`, outro se `False`
- Sempre executa um dos blocos

**Estrutura `if-elif-else`:**
- Permite múltiplas condições
- Verifica na ordem e para no primeiro `True`
- `else` é opcional

**Indentação:**
- É obrigatória em Python
- Define o que está dentro do bloco
- Use 4 espaços ou 1 tab

**Operadores Lógicos:**
- `and`: ambas condições verdadeiras
- `or`: pelo menos uma verdadeira
- `not`: inverte o valor

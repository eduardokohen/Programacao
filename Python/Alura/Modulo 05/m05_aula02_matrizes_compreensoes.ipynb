## **Resumo da Aula 2**

Nesta aula, você aprendeu sobre matrizes e compreensões, técnicas poderosas para trabalhar com dados.

### **Matrizes (listas de listas)**
- Matriz = lista que contém outras listas
- Acesso: `matriz[linha][coluna]`
- Útil para dados tabulares e grades
- Iteração: usar `enumerate` quando precisar de índices

```python
matriz = [[1, 2], [3, 4]]
valor = matriz[0][1]  # acessa linha 0, coluna 1
```

### **Compreensões de Listas**
- Sintaxe concisa para criar listas
- Formato base: `[expr for item in iterável]`
- Com filtro: `[expr for item in iterável if condição]`
- Aninhadas: `[[...] for i in range(n) for j in range(m)]`

```python
quadrados = [x**2 for x in range(5)]
positivos = [x for x in numeros if x > 0]
```

### **Compreensões de Dicionários**
- Similar às listas, mas cria dicionários
- Formato: `{chave: valor for ...}`
- Com filtro: `{chave: valor for ... if condição}`

```python
dicionario = {nome: idade for nome, idade in zip(nomes, idades)}
aprovados = {nome: nota for nome, nota in notas.items() if nota >= 7}
```

### **Dica importante:**
Se uma compreensão ficar difícil de ler, prefira usar `for` tradicional. Clareza > brevidade!

## **1. Estruturas de dados (visão rápida)**

Em Python, as estruturas mais usadas para organizar dados são:

- **Listas (`list`)**: coleção **ordenada** e **mutável**
- **Tuplas (`tuple`)**: coleção **ordenada** e **imutável**
- **Dicionários (`dict`)**: mapeamento **chave → valor** (ordem de inserção a partir do Python 3.7)

A ideia da aula é aprender o **mínimo necessário** para criar, acessar, atualizar e percorrer cada uma delas.

## **3. Tuplas (`tuple`) — quando você quer imutabilidade**

Tuplas são como listas, mas **não devem ser alteradas**.

Use quando:
- o conjunto de valores é “fixo” (ex.: coordenadas)
- você quer **garantia** de que não vai mudar sem querer

Acesso por índice e iteração funcionam igual às listas.

- `tupla[i]` acessa um item
- `a, b = tupla` desempacota (quantidade deve bater)

### 3.2 Iteração

Mesma ideia de lista:

- retornos múltiplos de função

## **4. Dicionários (`dict`) — chave → valor**

Use dicionários quando você quer acessar dados por uma **chave** (nomeada), em vez de índice.

- **Criar**: `{}` ou `dict()`
- **Acessar**: `d[chave]` (exige existir) ou `d.get(chave, padrão)`
- **Atualizar**: `d[chave] = valor`
- **Iterar**: `for chave, valor in d.items()`

```python
# Crie um dicionário pessoa = {"nome": "João", "idade": 25}
# Acesse o valor da chave "nome" usando pessoa["nome"] e imprima
# Acesse a chave "telefone" usando get() com valor padrão "(não informado)" e imprima
# Adicione a chave "cidade" com valor "São Paulo"
# Atualize a idade incrementando em 1
# Use um laço for com items() para iterar e imprimir cada chave e valor
# Remova a chave "cidade" usando pop() e imprima o dicionário final
# Escreva seu código aqui
```

### 4.1 Acesso seguro

- `d[chave]` é ótimo quando a chave **existe**
- `d.get(chave, padrão)` evita `KeyError` quando pode faltar

### 4.2 Atualizar e remover

- `d[chave] = valor` adiciona/atualiza
- `pop(chave)` remove e retorna o valor
- `del d[chave]` remove (sem retorno)

### 4.3 Iterar (a forma mais comum)

Use `items()` quando quiser chave e valor:

- `for chave, valor in d.items(): ...`

## **5. Comparação rápida**

| Característica | `list` | `tuple` | `dict` |
|---|---:|---:|---:|
| Mantém ordem? | ✅ | ✅ | ✅ (ordem de inserção) |
| Mutável? | ✅ | ❌ | ✅ |
| Acesso principal | índice | índice | chave |
| Duplicatas | ✅ | ✅ | ✅ (valores) / ❌ (chaves) |

## **Resumo da Aula 1**

Nesta aula, você aprendeu sobre as três principais estruturas de dados em Python:

### **Listas (`list`)**
- **Mutáveis** e **ordenadas**
- Usar quando precisa de sequências que podem crescer/encolher
- Operações: `append()`, `remove()`, `pop()`, `insert()`, `extend()`
- Acesso por índice: `lista[i]`, `lista[a:b]` (slicing)

### **Tuplas (`tuple`)**
- **Imutáveis** e **ordenadas**
- Usar quando precisa de dados fixos que não devem mudar
- Acesso igual às listas, mas não pode modificar elementos
- Útil para desempacotamento: `x, y = coordenadas`

### **Dicionários (`dict`)**
- **Mutáveis** e **ordenados** (ordem de inserção)
- Usar quando precisa acessar dados por **chave** (nomeada)
- Operações: `d[chave] = valor`, `d.get(chave, padrão)`, `pop()`
- Iteração comum: `for chave, valor in d.items()`

### **Escolha a estrutura certa:**
- **Lista**: sequências mutáveis (ex: lista de compras)
- **Tupla**: dados fixos (ex: coordenadas)
- **Dicionário**: acesso por nome/chave (ex: cadastro de pessoas)
